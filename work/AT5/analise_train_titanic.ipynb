As análises incluem:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Para exibir gráficos diretamente no notebook
%matplotlib inline
```

## 1. Carregando o dataset

O primeiro passo é carregar o arquivo `train.csv` com o `pandas`.

```python
df = pd.read_csv("train.csv")

# Visualizar as primeiras linhas do dataset
df.head()
```

## 2. Quantidade de linhas e colunas

A propriedade `.shape` mostra a dimensão do dataset. O primeiro valor representa a quantidade de linhas e o segundo representa a quantidade de colunas.

```python
linhas, colunas = df.shape

print(f"Quantidade de linhas: {linhas}")
print(f"Quantidade de colunas: {colunas}")
```

## 3. Tipo de dado de cada variável


```python
df.dtypes
```

## 4. Resumo estatístico das variáveis


```python
df.describe(include="all")
```

## 5. Tabela de frequência e gráfico da variável `Survived`

No dataset clássico do Titanic, a variável `Survived` indica:

- `0`: não sobreviveu;
- `1`: sobreviveu.

Como essa variável é categórica/binária, uma tabela de frequência ajuda a contar quantas pessoas estão em cada categoria.

```python
freq_survived = df["Survived"].value_counts().sort_index()
freq_survived
```

Também podemos calcular a frequência percentual.

```python
freq_survived_percentual = df["Survived"].value_counts(normalize=True).sort_index() * 100
freq_survived_percentual
```

### Gráfico da variável `Survived`

```python
freq_survived.plot(kind="bar")

plt.title("Frequência de Sobrevivência")
plt.xlabel("Sobreviveu")
plt.ylabel("Quantidade de passageiros")
plt.xticks([0, 1], ["Não sobreviveu", "Sobreviveu"], rotation=0)
plt.show()
```

## 6. Quantas pessoas sobreviveram?

Como `1` representa sobrevivente, a soma da coluna `Survived` indica o total de pessoas que sobreviveram.

```python
sobreviveram = df["Survived"].sum()

print(f"Quantidade de pessoas que sobreviveram: {sobreviveram}")
```

## 7. Qual a classe mais frequente?


```python
classe_mais_frequente = df["Pclass"].mode()[0]

print(f"Classe mais frequente: {classe_mais_frequente}")
```

Tabela de frequência das classes:

```python
freq_classe = df["Pclass"].value_counts().sort_index()
freq_classe
```

```python
freq_classe.plot(kind="bar")

plt.title("Frequência por Classe")
plt.xlabel("Classe")
plt.ylabel("Quantidade de passageiros")
plt.xticks(rotation=0)
plt.show()
```

## 8. Relação entre `Sex` e `Survived`

Para avaliar a relação entre sexo e sobrevivência, podemos usar uma tabela cruzada com `pd.crosstab()`.

```python
tabela_sex_survived = pd.crosstab(df["Sex"], df["Survived"])
tabela_sex_survived.columns = ["Não sobreviveu", "Sobreviveu"]

tabela_sex_survived
```

```python
tabela_percentual_sex_survived = pd.crosstab(
    df["Sex"],
    df["Survived"],
    normalize="index"
) * 100

tabela_percentual_sex_survived.columns = ["Não sobreviveu (%)", "Sobreviveu (%)"]
tabela_percentual_sex_survived
```

### Gráfico da relação entre sexo e sobrevivência

O gráfico de barras agrupadas é adequado porque estamos comparando duas variáveis categóricas: `Sex` e `Survived`.

```python
tabela_sex_survived.plot(kind="bar")

plt.title("Sobrevivência por Sexo")
plt.xlabel("Sexo")
plt.ylabel("Quantidade de passageiros")
plt.xticks(rotation=0)
plt.legend(title="Situação")
plt.show()
```

### Mulheres sobreviveram mais que homens?

Para responder melhor, podemos calcular a taxa média de sobrevivência por sexo. Como `Survived` usa `0` e `1`, a média representa a proporção de sobreviventes.

```python
taxa_sobrevivencia_por_sexo = df.groupby("Sex")["Survived"].mean() * 100

taxa_sobrevivencia_por_sexo
```

```python
for sexo, taxa in taxa_sobrevivencia_por_sexo.items():
    print(f"Taxa de sobrevivência de {sexo}: {taxa:.2f}%")

if taxa_sobrevivencia_por_sexo.get("female", 0) > taxa_sobrevivencia_por_sexo.get("male", 0):
    print("Conclusão: mulheres sobreviveram proporcionalmente mais que homens.")
else:
    print("Conclusão: homens sobreviveram proporcionalmente mais que mulheres.")
```

## 9. Distribuição das idades dos passageiros

A variável `Age` é numérica contínua. Por isso, um histograma é adequado para visualizar sua distribuição.

Antes de criar o gráfico, verificamos quantos valores ausentes existem na coluna.

```python
valores_ausentes_age = df["Age"].isnull().sum()

print(f"Valores ausentes em Age: {valores_ausentes_age}")
```

### Histograma da idade

O histograma mostra a frequência de passageiros em intervalos de idade.

```python
df["Age"].dropna().plot(kind="hist", bins=20, edgecolor="black")

plt.title("Distribuição das Idades dos Passageiros")
plt.xlabel("Idade")
plt.ylabel("Frequência")
plt.show()
```

Também podemos usar medidas como média, mediana e assimetria para interpretar melhor a distribuição.

```python
media_idade = df["Age"].mean()
mediana_idade = df["Age"].median()
assimetria_idade = df["Age"].skew()

print(f"Média da idade: {media_idade:.2f}")
print(f"Mediana da idade: {mediana_idade:.2f}")
print(f"Assimetria da idade: {assimetria_idade:.2f}")
```

## 10. A maioria dos passageiros era jovem ou idosa?

Podemos criar faixas etárias para facilitar a interpretação.

```python
faixas = [0, 12, 18, 35, 60, 100]
rotulos = ["Criança", "Adolescente", "Jovem adulto", "Adulto", "Idoso"]

df["Faixa_Etaria"] = pd.cut(df["Age"], bins=faixas, labels=rotulos)

freq_faixa_etaria = df["Faixa_Etaria"].value_counts().sort_index()
freq_faixa_etaria
```

```python
freq_faixa_etaria.plot(kind="bar")

plt.title("Distribuição por Faixa Etária")
plt.xlabel("Faixa etária")
plt.ylabel("Quantidade de passageiros")
plt.xticks(rotation=45)
plt.show()
```
